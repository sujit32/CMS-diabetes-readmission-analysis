```python
df = df.drop(columns = 'weight')
```

```python
df['race'] = df['race'].replace('?', 'N/A')
```

```python
df['payer_code'] = df['payer_code'].replace('?', 'N/A')
```

```python
df['medical_specialty'] = df['medical_specialty'].replace('?', 'N/A')
```

```python
df['diag_1'] = df['diag_1'].replace('?', 'N/A')
```

```python
df['diag_2'] = df['diag_2'].replace('?', 'N/A')
```

```python
df['diag_3'] = df['diag_3'].replace('?', 'N/A')
```

```python
df['max_glu_serum'] = df['max_glu_serum'].fillna('Not Tested')
```

```python
df['A1Cresult'] = df['A1Cresult'].fillna('Not Tested')
```

```python
df["time_in_hospital"] = df["time_in_hospital"].astype(int)
```

```python
df['new_age'] = df['age'].str.replace(r'[\[\)]', '', regex = True)
```

```python
df['new_age'].head()
```

```python
df['new_age'] = (
    df['age']
    .str.replace("[", "", regex=False)
    .str.replace(")", "", regex=False)
    .str.split("-")
    .apply(lambda x: (int(x[0]) + int(x[1])) / 2)
)
```

```python
df['new_age']= df['new_age'].astype(int)
```
